```python
import pandas as pd

df = pd.read_csv('results.csv')

def calculate_mape_group(group):
    y_true = group['y_true']
    y_pred = group['y_pred']
    return 100 * (abs((y_true - y_pred) / y_true)).mean()

mape_result = df.groupby(['set_name', 'method_name']).apply(calculate_mape_group).reset_index(name='mape')
mean_time = df.groupby(['set_name', 'method_name'])['time'].mean().reset_index(name='mean_time')
result = pd.merge(mape_result, mean_time, on=['set_name', 'method_name'])

result
```
